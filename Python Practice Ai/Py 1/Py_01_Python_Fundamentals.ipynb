## 2. Variables

A variable is a name used to store a value.

Example:

name = "Sushant"

Here:
- `name` is the variable.
- `"Sushant"` is the value stored in the variable.

Python allows us to store and reuse data using variables.

```python
name = "Sushant"
age = 20

print(name)
print(age)
```

## 2.2 Creating Variables

```python

name = "Sushant"
age = 20
course = "B.Tech Computer Science"

print("Name:", name)
print("Age:", age)
print("Course:", course)

# Basic arithmetic
a = 10
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```
